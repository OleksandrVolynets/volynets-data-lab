```python
# Завдання 8.3. Створіть копію таблиці й заповніть у ній порожні значення amount медіаною цієї ж колонки. Виведіть отриману медіану і перевірте, що порожніх значень більше не лишилося.
orders_copy = orders.copy() # створюю копію DF
median_amount = orders_copy['amount'].median() # розраховую медіанне значення по стовпцю amount
orders_copy['amount'] = orders_copy['amount'].fillna(median_amount) # замінюю пусті значення в стовпці amount на медіанне
print(f"median: {median_amount}\n\n{orders_copy}") # вивожу медіанне значення та DF з заміненними значеннями
```

```python
# Завдання 8.4. Створіть ще одну таблицю, з якої видалені рядки з порожнім amount. Виведіть її розміри.
orders_clean = orders.dropna(subset=['amount']) # створюю новий DF, куди записую рядки з orders, в яких, в полі amount є значення
print(orders_clean.shape)
```

```python
# Завдання 9.1. Створіть копію таблиці у змінну work і збільште в ній усі значення amount на 50. Виведіть значення amount для ord_1 в обох таблицях і переконайтеся, що оригінал не змінився.
work = orders.copy() # створюю копію DF
work['amount'] = work['amount'] + 50 # додаю до кожного значення в полі amount 50

print(f" orders amount ord_1 = {orders.loc['ord_1', 'amount']}\n work amount ord_1 = {work.loc['ord_1', 'amount']}")
```

```python
# Завдання 9.2. Додайте в work нову колонку avg_item — середня вартість одного товару в замовленні, тобто amount поділена на items, округлена до двох знаків.
work['avg_item'] =work['amount']/work['items'].round(2) # додаю нову колонку в work зі значенням вартості 1 товару
```

```python
# Завдання 9.3. Перейменуйте в work колонку source на channel.
work = work.rename(columns={'source': 'channel'}) # перейменування колонки source у channel
```

```python
# Завдання 9.4. Видаліть із work колонку express і виведіть список колонок, що залишилися.
work = work.drop(columns=['express'], axis=1) # видаляю колонку express (axis=1 - колонка)
print(work.columns.tolist()) # список колонок, які залишились
```
